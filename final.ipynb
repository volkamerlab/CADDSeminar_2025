# 1 The Big Question: Why Do Some Drugs Dissolve and Others Don't?

Imagine dropping sugar in water, it vanishes instantly. Now try the same with olive oil, it just floats there. This is **aqueous solubility**: a molecule's ability to "make friends" with water [1].

For drugs, this is very important. A medicine can't work if it won't dissolve in your bloodstream! In fact, **40% of new drugs fail** simply because they're too "water shy" (hydrophobic) [2].

## The Chemistry Rule: "Like Dissolves Like"

Water is a polar molecule, think of it as having a positive end and negative end, like a tiny magnet. Molecules that are also polar (have charged or hydrogen bonding groups) love water and dissolve easily. Molecules that are greasy or oily (hydrophobic) avoid water like oil avoids vinegar in a salad dressing [3].

## The Chemistry Behind Dissolution: Five Key Drivers

From a chemistry lens, solubility is a **tug-of-war** between molecule-water attractions and molecule-molecule attractions in the solid. Five molecular properties dominate [1]:

### 1. Lipophilicity (Grease vs. Water-Loving)
The famous **logP** (octanol water partition) tells us if a molecule prefers organic (greasy) or aqueous environments. High logP = loves oil, hates water = poor solubility. This is our **strongest predictor**—features like SLogP should scream at our model [3].

### 2. Polarity & Hydrogen Bonding
Polar groups (**OH, NH, C=O**) love water because they form hydrogen bonds with H₂O molecules. More H-bond donors and acceptors = more favorable water interactions = better solubility. Think of **sugar** (tons of OH groups) dissolving instantly vs. **wax** (zero polar groups) floating [4].

### 3. Molecular Size
Bigger molecules need more energy to shove water molecules aside and create space. **Molecular weight** and bulkiness negatively correlate with solubility small = soluble, huge = not so much [4].

### 4. Ionization
Acids (**COOH**) and bases (**amines**) can gain or lose protons in water, becoming charged ions. Charged molecules love water (**charge-dipole interactions**). Count of acidic/basic groups should boost solubility [1].

### 5. Aromaticity
Flat aromatic rings (benzene, naphthalene) are hydrophobic and prefer stacking with themselves (**π-π stacking**) rather than mingling with water. More aromatic rings = lower solubility [2].

## Methods to Determine Aqueous Solubility of Compounds

### 1. Shake-Flask Method (Thermodynamic Solubility) [5]
**Gold standard** for equilibrium solubility of crystalline solids.

**Protocol**: Excess pure compound powder → aqueous buffer (pH 7.4, 25°C) → agitated 24-72h → filtered (0.2 μm) → HPLC-UV/LC-MS quantification.  
**Accuracy**: ±0.3 logS units.  
**Limitation**: Time intensive (days/compound).  
**Use**: Development stage confirmation.

### 2. Turbidimetric/Nephelometric Method (Kinetic Solubility) [5]
**High-throughput** screening from DMSO stock (1-10 mM).

**Protocol**: Serial dilutions → aqueous buffer → laser light scattering (90° detection) monitors precipitation.  
**Throughput**: Thousands/day.  
**Limitation**: Measures amorphous precipitate (overestimates crystalline by **1-2 logS**).  
**Use**: Early discovery.

### 3. Computational Prediction [4]
Structure-based tools (**SwissADME, ChemAxon, ACD/Solubility**) estimate logS from molecular descriptors.  
**Accuracy**: ±0.8-1.2 logS.  
**Best for**: Relative ranking, not absolute values.

## Limitations in Obtaining Aqueous Solubility Data in the Lab

1. **Time Bottleneck** (Shake-Flask): 24-72h for equilibrium → ~20 compounds/day [5]
2. **Material Scarcity**: Needs 1-5 mg pure crystalline powder [5]
3. **High Variability (±0.5 logS)**: pH drift, temp fluctuations (±1°C = 0.1 logS), polymorph changes [6]
4. **Solid-State Chaos**: Metastable → stable forms (amorphous vs crystalline: **1-2 logS diff**) [6]
5. **Turbidimetry Pitfalls**: DMSO boosts solubility; sensitive to bubbles/dust [5]

## AqSol Lab Measurement: Risks & Prevention Strategies

Aqueous solubility (**AqSol**) faces challenges compromising reliability [6].

| **What Can Go Wrong** | **Prevention Strategies** |
|-----------------------|---------------------------|
| **Insufficient Equilibration** (>24h = kinetic overestimate by 1+ logS) [5] | Shake-Flask: **72h agitation + 24h settling** |
| **Phase Transformation** (metastable → stable drop) [6] | **Excess solid (10-20x)** + pre/post-XRD |
| **pH/Temp Drift** (pH ±0.2 = ±0.5 logS; +1°C = +0.1 logS) [6] | **Automated controls** (±0.1°C, ±0.05 pH) |
| **Material Issues** (floaty powders, DMSO artifacts) [5] | Turbidimetry: **<1% DMSO** + triplicates + blanks |
| **Filtration Errors** (adsorption/underestimate) [5] | **Multiple filters tested** |

**General Best Practices**:
- >98% pure material + polymorph characterization
- Standardized protocols (USP, OECD)
- **Uncertainty reporting** mandatory

**Recommendation**: Prioritize **relative ranking** over absolute logS. Confirm kinetic with thermodynamic [5].

## Conclusion

Aqueous solubility measurement remains a **cornerstone challenge** despite refinement. Tension between **thermodynamic accuracy** (slow shake-flask) vs. **throughput** (fast turbidimetry) creates unavoidable trade-offs [5].

### Key Takeaways:
- **Risks inevitable**: ±0.3-0.5 logS uncertainty guaranteed [6]
- **Mitigation works**: 72h protocols + controls **reduce** errors [5]
- **Reality**: Even perfect labs vary; inter-facility = **0.5-1 logS spread** [6]
- **Practical**: Kinetic triage → thermodynamic decisions. **logS > -6** oral target [2]

**Bottom line**: AqSol reflects experimental reality **noisy but actionable** with controls. Chemistry demands skepticism of single point "truths" [6].

### 2.1 Premise — What Data Should Look Like for Aqueous Solubility Modeling
The objective of this project is to predict aqueous solubility, generally described as log10 solubility (LogS, mol/L), directly from molecular structure encoded as SMILES. SMILES is a compact and widely accepted representation of chemical graphs for computational modeling (Weininger, 1988).
Importantly, aqueous solubility is not a single intrinsic constant of a molecule. Reported values depend on how the experiment is defined and performed, including whether solubility is measured under thermodynamic (equilibrium) or kinetic (apparent) conditions (Llompart et al., 2024). Even within classical shake-flask experiments, solubility depends on pH, temperature, buffer composition, cosolvents, solid form, equilibration time, and analytical method (Wenlock et al., 2011). Replicate measurements and uncertainty estimates are also important because aqueous solubility measurements are known to exhibit substantial experimental variability, even for the same compound measured under similar conditions (Saal et al., 2012)
In chemical databases, aqueous solubility values for small organic molecules cover a very broad range, from highly soluble compounds (around LogS +1) to extremely poorly soluble ones (around LogS −10) (Wenlock et al., 2011; Llompart et al., 2024).
As a pharmaceutical R&D team, we are not only interested in predicting aqueous solubility as a numerical property, but in understanding whether a molecule’s solubility falls within a range that is meaningful for oral drug development. From this perspective, solubility values above approximately LogS −3 are generally considered favorable, while values below roughly LogS −6 often indicate severe formulation challenges. These solubility ranges represent empirical guidelines commonly used in pharmaceutical research rather than strict thresholds. The most informative region for pharmaceutical modeling lies between these boundaries, where solubility constraints exist but do not yet rule out oral development (Wenlock et al., 2011; Saal et al., 2012; Llompart et al., 2024).

### 2.2 Ideal Dataset Definition for Pharmaceutical Solubility Modeling
From this perspective, an ideal dataset for solubility modeling would include:
• X — Structure: canonical SMILES with consistent handling of ionization states and salts
• Y — Experimental context: pH, temperature, buffer composition, cosolvent fraction, and assay type
• Z — Measurement definition & reliability: clear distinction between kinetic and equilibrium solubility, together with replicate measurements or uncertainty estimates
In addition, an ideal dataset should satisfactorily cover the solubility range that is most relevant for pharmaceutical decision-making, particularly the intermediate region (approximately LogS −3 to −6), while still including sufficiently soluble and extremely insoluble compounds to capture the full solubility landscape.
Such a dataset would also approximate independent and identically distributed (i.i.d.) samples, thereby minimizing hidden dataset shift between training and evaluation data (Quionero-Candela et al., 2009; Rabanser et al., 2019).

When choosing how to represent molecules, it is important to think about the solubility range that each dataset mainly contains. During model training, it should be kept in mind that performance is unlikely to be uniform across all solubility ranges. Models trained on broad datasets such as AqSolDB may perform well on average but fail to capture behavior in pharmaceutically relevant regions. ## **3.1 Sanitization and Standardization**

### **Molecular Sanitization**

Raw SMILES strings may contain invalid valences, incorrect aromaticity, or non-chemical syntax, so we sanitize each molecule to ensure valid chemistry and canonicalize SMILES for consistent representation before feature generation.

```python
# Sanitize SMILES and keep only chemically valid molecules
valid_smiles = []
valid_logs = []

for _, row in df.iterrows():
    smi = row['SMILES']
    logS = row['LogS']
    mol = Chem.MolFromSmiles(smi)
    if mol is None:
        continue
    try:
        # Check valence, aromaticity, etc., and canonicalize
        SanitizeMol(mol)
        clean_smi = Chem.MolToSmiles(mol, canonical=True)
        valid_smiles.append(clean_smi)
        valid_logs.append(logS)
    except Exception:
        # Skip molecules that fail sanitization
        continue

clean_df = pd.DataFrame({'SMILES': valid_smiles, 'LogS': valid_logs})
print(f" Valid molecules after sanitization: {len(clean_df)} / {len(df)}")
clean_df.head()
```

Sanitization retained 9,980 of 9,982 molecules (99.98%). The two dropped entries represent chemically invalid structures that cannot be converted to meaningful descriptors, and removing them prevents errors during feature generation while preserving nearly the entire dataset.

### **Molecular Standardization**

Solubility can vary with salt forms, counterions, or non-standard representations. We standardize each molecule to remove salts, normalize functional groups, and reionize to a consistent form, to ensure that we model the organic component's intrinsic solubility rather than artifacts of salt form.

```python
# Standardization (remove salts, normalize, reionize)
def standardize_molecule(mol):
    try:
        cleaner = rdMolStandardize.Cleanup(mol)
        normalizer = rdMolStandardize.Normalize(cleaner)
        reionizer = rdMolStandardize.Reionize(normalizer)
        return reionizer
    except:
        return mol

# Standardize entire dataset
standardized_smiles = []
for smi in df['SMILES']:
    mol = Chem.MolFromSmiles(smi)
    if mol is not None:
        mol = standardize_molecule(mol)
        clean_smi = Chem.MolToSmiles(mol, canonical=True)
        standardized_smiles.append(clean_smi)
    else:
        standardized_smiles.append(None)

df['Standardized_SMILES'] = standardized_smiles
df = df.dropna(subset=['Standardized_SMILES']).reset_index(drop=True)

df.drop(columns="Unnamed: 0", inplace=True, errors="ignore")
df = df[['SMILES', 'Standardized_SMILES', 'LogS']]

df.to_csv("./data/AqSolDB_Standardized.csv", index=False)
df = pd.read_csv("./data/AqSolDB_Standardized.csv")
df.head()
```



In this notebook we use two complementary paradigms generated from SMILES: **descriptor-based Mordred features** and **fingerprint-based MACCS keys**.

### **Mordred Descriptors**

Mordred [2], a freely available descriptor calculator, can compute 1,800+ molecular features from SMILES, including 2D and 3D descriptors.  

These include *simple composition counts (atoms, bonds)*, *key physicochemical properties* such as polarity, hydrogen‑bonding capacity, and lipophilicity, *3D desciptors*, and many more.

### **MACCS Keys**  

MACCS [3][4] (Molecular ACCess System) keys are a type of 2D structural molecular fingerprint that represent each molecule as a 166-bit binary string, where each bit indicates the presence or absence of one of the 166 predefined substructural features.  

They capture patterns such as *aromatic rings*, *functional groups*, and *atom types* in a very compact way, which makes them a good contrast to the richer, continuous Mordred descriptors.  
  
  
Together, they allow us to capture detailed physicochemical properties and key structural motifs of the molecules.

```python
def generate_features(train_df, test_df, descriptors):
    """
    Generate raw Mordred descriptors and MACCS fingerprints from SMILES.

    Parameters:

    train_df, test_df : pd.DataFrame
        DataFrames from Original Data containing 'Standardized_SMILES' and 'LogS' columns.
    descriptors : module or list
        Mordred descriptor module (or subset) to compute.

    Returns:

    X_train_mordred_raw, X_test_mordred_raw : pd.DataFrame
        Raw Mordred descriptor matrices (numeric).
    X_train_maccs_raw, X_test_maccs_raw : np.ndarray
        MACCS fingerprint matrices (shape [n_samples, 166]).
    y_train, y_test : np.ndarray
        LogS target values for training and testing.
    """

    print("Generating molecular representations...")
    mols_train = [Chem.MolFromSmiles(smi) for smi in train_df["Standardized_SMILES"]]
    mols_test  = [Chem.MolFromSmiles(smi) for smi in test_df["Standardized_SMILES"]]

    # Mordred Descriptors
    print("\n1. Mordred Descriptors (2D only)...")

    # Compute 2D Mordred descriptors for train and test sets
    calc = mordred.Calculator(descriptors, ignore_3D=True)
    X_train_mordred_raw = calc.pandas(mols=mols_train)
    X_test_mordred_raw  = calc.pandas(mols=mols_test)

    # Convert all descriptor columns to numeric (coerce invalid to NaN)
    X_train_mordred_raw = X_train_mordred_raw.apply(pd.to_numeric, errors='coerce')
    X_test_mordred_raw  = X_test_mordred_raw.apply(pd.to_numeric, errors='coerce')

    print(f"   Raw shapes: Train {X_train_mordred_raw.shape}, Test {X_test_mordred_raw.shape}")

    # MACCS Keys
    print("\n2. MACCS Keys...")
    def maccs_fp(mol):
          fp = rdMolDescriptors.GetMACCSKeysFingerprint(mol)
          return np.array(fp.ToList(), dtype=np.uint8)

    # Generate MACCS fingerprints
    X_train_maccs_raw = np.array([maccs_fp(mol) for mol in mols_train])
    X_test_maccs_raw  = np.array([maccs_fp(mol) for mol in mols_test])

    print(f"   Raw shapes: Train {X_train_maccs_raw.shape}, Test {X_test_maccs_raw.shape}")

    # Common targets
    y_train = train_df['LogS'].values
    y_test  = test_df['LogS'].values

    print("\nMolecular Features Generated: Mordred and MACCS.")
    return (X_train_mordred_raw, X_test_mordred_raw,
            X_train_maccs_raw, X_test_maccs_raw,
            y_train, y_test)
```

```python
# Generating features

raw_features = generate_features(train_df, test_df, descriptors)
```

### **3.3 Feature Cleaning**

Based on our inspection:

**Mordred descriptors** showed NaN values, which might be from failed calculations. Thus, feature cleaning is required, where we will perform NaN removal, zero-variance filtering, low-variance pruning, and correlation-based deduplication.

We apply a sequential filtering pipeline where **column selection is based solely on the training set** to prevent data leakage. The same columns are then dropped from the test set to maintain feature alignment:

- **Remove `FilterItLogS`**, which is a built-in solubility predictor in Mordred, and retaining it would leak the target variable into features. [[PMC10583449](https://pmc.ncbi.nlm.nih.gov/articles/PMC10583449/)]

- **Drop NaN-containing columns**, identified using training data only. Test set columns are dropped to match.

- **Drop zero-variance features**, descriptors constant across all training molecules carry no predictive signal.

- **Prune bottom 5% by variance**, removes near-constant descriptors that add noise without meaningful signal.

- **Remove highly correlated pairs (>0.95)**, eliminates redundancy to improve model stability.

This reduces dimensionality, ensuring interpretability and preserving chemically meaningful variation.  

```python
# Mordred Feature Cleaning

def clean_mordred_features(X_train_mordred_raw, X_test_mordred_raw):
    """
    Clean Mordred descriptors using train set statistics only to prevent data leakage.

    Applies sequential filtering:
      (1) remove FilterItLogS target leak,
      (2) drop NaN columns,
      (3) remove zero-variance features,
      (4) prune bottom 5% by variance,
      (5) remove highly correlated pairs (>0.95).
    Same columns dropped from test set for alignment.

    Returns cleaned train-test set with reduced dimensionality.
    """
    print("\nCleaning mordred features...")

    print(f"  Initial shapes - Train: {X_train_mordred_raw.shape}, Test: {X_test_mordred_raw.shape}")

    # Remove 'FilterItLogS' to avoid data leakage during test
    # see data preparation section: https://pmc.ncbi.nlm.nih.gov/articles/PMC10583449/#:~:text=We%20used%20Mordred%20package%20%5B23,a%20selection%20of%20506%20descriptors.
    print("  0. Removing 'FilterItLogS' to avoid data leakage...", end=" ")

    X_train_mordred = X_train_mordred_raw.drop(columns=['FilterItLogS'], errors='ignore')
    X_test_mordred  = X_test_mordred_raw.drop(columns=['FilterItLogS'], errors='ignore')
    print(f" Train: {X_train_mordred_raw.shape[1]} to {X_train_mordred.shape[1]} cols")

    # 1. Drop descriptor columns that contain any NaN values
    print("  1. Removing features with NaN values...", end=" ")

    nan_cols = X_train_mordred.columns[X_train_mordred.isna().any()].tolist()
    n_dropped1 = len(nan_cols)

    X_train_mordred = X_train_mordred.drop(columns=nan_cols)
    X_test_mordred  = X_test_mordred.drop(columns=nan_cols)

    print(f"dropped {n_dropped1}; {X_train_mordred.shape[1]} features")

    # 2. Drop descriptors that are zero for all molecules
    print("  2. Removing features with zero values...", end=" ")

    zero_cols = X_train_mordred.columns[(X_train_mordred == 0).all()]
    n_dropped2 = len(zero_cols)

    X_train_mordred = X_train_mordred.drop(columns=zero_cols)
    X_test_mordred  = X_test_mordred.drop(columns=zero_cols)

    print(f"dropped {n_dropped2}; {X_train_mordred.shape[1]} features")

    # 3. Drop descriptors with low variance (bottom 5%)
    print("  3. Removing low variance...", end=" ")

    var_thresh = X_train_mordred.var().nsmallest(int(0.05*X_train_mordred.shape[1])).index
    n_dropped3 = len(var_thresh)

    X_train_mordred = X_train_mordred.drop(columns=var_thresh)
    X_test_mordred  = X_test_mordred.drop(columns=var_thresh)

    print(f"dropped {n_dropped3}; {X_train_mordred.shape[1]} features")

    # 4. Drop highly correlated descriptors
    print("  4. Removing correlated features...", end=" ")

    corr_matrix = X_train_mordred.corr().abs()

    # Upper triangle (avoid self-correlation)
    upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

    # Find highly correlated features (>0.95 threshold)
    threshold = 0.95
    to_drop = [column for column in upper.columns if any(upper[column] > threshold)]

    n_dropped4 = len(to_drop)

    X_train_mordred_final = X_train_mordred.drop(columns=to_drop)
    X_test_mordred_final  = X_test_mordred.drop(columns=to_drop)

    print(f"dropped {n_dropped4}; Final {X_train_mordred_final.shape[1]} features")

    reduction = 100 * X_train_mordred_final.shape[1] / X_train_mordred_raw.shape[1]
    print(f"   Result: {X_train_mordred_final.shape[1]} features ({reduction:.0f}% of original)")

    return X_train_mordred_final, X_test_mordred_final
```

```python
def add_features_to_data(train_df, test_df,
                        X_train_mordred, X_test_mordred,
                        X_train_maccs, X_test_maccs,
                        output_dir="./data/"):
    """
    Combines Mordred to dataset and  MACCS to dataset.

    Args:
        train_df, test_df (pd.DataFrame): Original data with 'Standardized_SMILES', 'LogS'
        X_train_mordred, X_test_mordred: Cleaned Mordred descriptors
        X_train_maccs, X_test_maccs: Cleaned MACCS fingerprints
    """
    import os

    # Output directory
    os.makedirs(output_dir, exist_ok=True)

    # standardize input formats

    def ensure_dataframe(X_train, X_test, feature_name, n_train, n_test):
        """
        Convert numpy arrays to DataFrames with descriptive column names
        """
        if isinstance(X_train, np.ndarray):
            X_train = pd.DataFrame(X_train,
                                 columns=[f"{feature_name}_{i}" for i in range(X_train.shape[1])])
        if isinstance(X_test, np.ndarray):
            X_test = pd.DataFrame(X_test,
                                columns=[f"{feature_name}_{i}" for i in range(X_test.shape[1])])
        return X_train.reset_index(drop=True), X_test.reset_index(drop=True)

    n_train, n_test = len(train_df), len(test_df)
    X_train_mordred, X_test_mordred = ensure_dataframe(X_train_mordred, X_test_mordred,
                                                      "Mordred", n_train, n_test)
    X_train_maccs, X_test_maccs = ensure_dataframe(X_train_maccs, X_test_maccs,
                                                  "MACCS", n_train, n_test)

    print(f"Mordred: {X_train_mordred.shape[1]} features")
    print(f"MACCS:  {X_train_maccs.shape[1]} features")

    # Prepare SMILES and LogS
    smiles_logs_train = train_df[["Standardized_SMILES", "LogS"]].reset_index(drop=True)
    smiles_logs_test = test_df[["Standardized_SMILES", "LogS"]].reset_index(drop=True)

    # Save 4 datasets

    # Train-Mordred
    train_mordred = pd.concat([smiles_logs_train, X_train_mordred], axis=1)
    train_mordred.to_csv(os.path.join(output_dir, "solubility_train_mordred.csv"), index=False)

    # Test-Mordred
    test_mordred = pd.concat([smiles_logs_test, X_test_mordred], axis=1)
    test_mordred.to_csv(os.path.join(output_dir, "solubility_test_mordred.csv"), index=False)

    # Train-MACCS
    train_maccs = pd.concat([smiles_logs_train, X_train_maccs], axis=1)
    train_maccs.to_csv(os.path.join(output_dir, "solubility_train_maccs.csv"), index=False)

    # Test-MACCS
    test_maccs = pd.concat([smiles_logs_test, X_test_maccs], axis=1)
    test_maccs.to_csv(os.path.join(output_dir, "solubility_test_maccs.csv"), index=False)


    print("\nFiles saved.")
    print(f" Mordred Train: solubility_train_mordred.csv  {train_mordred.shape}")
    print(f" Mordred Test:  solubility_test_mordred.csv   {test_mordred.shape}")
    print(f" MACCS Train:  solubility_train_maccs.csv   {train_maccs.shape}")
    print(f" MACCS Test:   solubility_test_maccs.csv    {test_maccs.shape}")

    return {
        # Mordred datasets
        'train_mordred_csv': os.path.join(output_dir, "solubility_train_mordred.csv"),
        'test_mordred_csv': os.path.join(output_dir, "solubility_test_mordred.csv"),
        'train_mordred_df': train_mordred,
        'test_mordred_df': test_mordred,

        # MACCS datasets
        'train_maccs_csv': os.path.join(output_dir, "solubility_train_maccs.csv"),
        'test_maccs_csv': os.path.join(output_dir, "solubility_test_maccs.csv"),
        'train_maccs_df': train_maccs,
        'test_maccs_df': test_maccs
    }
```

## Modeling (Theory) — ML pipeline, SVR/RF mechanics, CV tuning, and evaluation for LogS

### Overview
In this section, we model aqueous solubility as a **supervised regression** problem: we train models that map a molecule (encoded as a fixed-length feature vector) to a continuous solubility label (**LogS**). AqSolDB aggregates curated solubility measurements from multiple sources, so the target distribution includes experimental noise and heterogeneity. This motivates robust baselines and careful validation. ([AqSolDB](https://doi.org/10.1038/s41597-019-0151-1))

### Inputs and outputs
**Inputs (X): fixed-length numeric feature vectors**
- **Mordred descriptors:** high-dimensional continuous descriptors computed from 2D molecular structure. ([Mordred](https://doi.org/10.1186/s13321-018-0258-y))
- **MACCS features:** binary (0/1) substructure indicators stored as a fixed set of feature columns in our dataset schema. (Implementation: [RDKit MACCS keys](https://www.rdkit.org/new_docs/source/rdkit.Chem.MACCSkeys.html))

**Output (y):**
- **LogS** (continuous real value). Because the target is continuous, the task is **regression**, independent of whether the input features are binary (MACCS) or continuous (Mordred).

### ML workflow (what the code does)
1. Load AqSolDB train/test feature matrices (tabular `X` and `y`).
2. For Challenge 1/2 (SMILES + LogS), generate the same feature type (Mordred or MACCS) and **align columns** to the training schema so the models receive identical feature definitions.
3. Fit SVR and RF using pipelines for correct preprocessing.
4. Select hyperparameters using **5-fold cross-validation** with grid search.
5. Evaluate on AqSolDB test (in-distribution) and Challenge 1/2 (external generalization).

---

### Why the data structure fits SVR and RF
SVR and RF assume:
- a **numeric feature matrix** with a fixed set of columns (one row per molecule), and
- a **continuous target** (one value per molecule).

Mordred and MACCS satisfy this by providing fixed-length vectors for each molecule. These models do not operate on SMILES directly; SMILES must be transformed into descriptors/fingerprints first.

---

### Models and mechanics

#### Support Vector Regression (SVR) with RBF kernel
SVR is a regularized regression method. With a non-linear **RBF kernel**, it can model complex structure–property relationships by using similarity in feature space. ([SVR tutorial](https://doi.org/10.1023/B:STCO.0000035301.49549.88))

**SVR prediction logic (inference)**
After training, SVR retains a subset of influential training samples called **support vectors**.  
To predict LogS for a new molecule, SVR:
1) computes similarity between the new molecule and each support vector in feature space (RBF kernel),  
2) assigns higher influence to more similar support vectors, and  
3) outputs a weighted combination of these influences plus an intercept term.  
The hyperparameters (`C`, `epsilon`, `gamma`) control how many points become support vectors and how local/smooth the learned mapping is.  
Reference: [Smola & Schölkopf, 2004](https://doi.org/10.1023/B:STCO.0000035301.49549.88)

**Why preprocessing matters for SVR**
SVR with RBF is distance-based, so feature scales affect similarity. Therefore we use:
- **Scaling (StandardScaler)** so features contribute comparably to distance computations.
- **Median imputation** for missing values (when present), fit on training data only to avoid leakage.  
(Workflow: [scikit-learn Pipeline](https://scikit-learn.org/stable/modules/compose.html#pipeline))

**SVR hyperparameters (interpretation)**
- `C`: trade-off between training fit and regularization (higher values fit more tightly; lower values increase smoothness).
- `gamma`: locality of the RBF effect (higher values allow more local complexity; lower values enforce smoother behavior).
- `epsilon`: tolerance band that reduces sensitivity to small deviations consistent with experimental noise.

**MACCS + SVR note (similarity geometry)**
MACCS features are binary. Regression remains valid, but the RBF kernel uses Euclidean geometry, whereas fingerprint similarity in cheminformatics is often quantified using **Tanimoto/Jaccard overlap** (shared present bits divided by total present bits). This measure focuses on overlap of present substructures and ignores shared zeros, which is appropriate for sparse fingerprints. Thus MACCS+RBF-SVR is a valid baseline but may be limited by a representation–kernel mismatch. ([RDKit MACCS keys](https://www.rdkit.org/new_docs/source/rdkit.Chem.MACCSkeys.html))

#### Random Forest Regression (RF)
Random Forest averages predictions from many decision trees trained on bootstrapped samples and randomized feature subsets. Averaging reduces variance and yields a robust non-linear baseline. ([Breiman, 2001](https://doi.org/10.1023/A:1010933404324))

**RF prediction logic (inference)**
A Random Forest consists of many decision trees. Each tree applies a sequence of threshold-based splits on individual features.  
To predict LogS for a new molecule, RF:
1) sends the feature vector through each tree,  
2) each tree routes the molecule to a terminal **leaf** that stores a numeric LogS estimate (typically the mean LogS of training samples in that leaf),  
3) the forest outputs the **average** of all tree predictions.  
Reference: [Breiman, 2001](https://doi.org/10.1023/A:1010933404324)

**Why RF is suitable here**
- Captures non-linearities and feature interactions in tabular descriptors.
- Does not require scaling (splits are threshold-based).
- Variance reduction via ensembling helps under noisy experimental targets such as solubility. ([Breiman, 2001](https://doi.org/10.1023/A:1010933404324))

**RF hyperparameters (interpretation)**
- `n_estimators`: number of trees (more trees generally increases stability).
- `max_depth`: maximum tree depth (controls complexity/overfitting).
- `min_samples_leaf` / `min_samples_split`: prevent overly specific splits (regularization).
- `max_features`: feature subsampling that promotes tree diversity and can improve generalization.

---

### Hyperparameter selection with cross-validation (CV)
We select hyperparameters using **5-fold cross-validation** on the training split with **GridSearchCV**, optimizing RMSE (via `neg_root_mean_squared_error`). This reduces dependence on a single split and provides a principled way to control model complexity. ([GridSearchCV](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GridSearchCV.html))

---

### Evaluation metrics

We report:
- **RMSE**: penalizes large errors more strongly (used for CV selection).
- **MAE**: average absolute error (more robust to outliers than RMSE).
- **R²**: performance relative to predicting the dataset mean; can be negative under strong dataset shift.
- **CCC (Concordance Correlation Coefficient)**: agreement with the identity line (captures correlation and calibration; useful when external shift causes systematic bias). ([Lin, 1989](https://doi.org/10.2307/2532051))

```python
# ===== Load datasets =====
train_mord = pd.read_csv(TRAIN_MORDRED)
test_mord  = pd.read_csv(TEST_MORDRED)

train_macc = pd.read_csv(TRAIN_MACCS)
test_macc  = pd.read_csv(TEST_MACCS)

ch1 = pd.read_csv(CH1_PATH)
ch2 = pd.read_csv(CH2_PATH)

print("Train Mordred:", train_mord.shape, "| Test Mordred:", test_mord.shape)
print("Train MACCS:  ", train_macc.shape, "| Test MACCS:  ", test_macc.shape)
print("Challenge 1:  ", ch1.shape, "| Challenge 2:  ", ch2.shape)

TARGET_COL = "LogS"
ID_COL = "Standardized_SMILES"  # present in train/test descriptor CSVs
CH_SMILES_COL = "SMILES"        # present in challenge CSVs
```

### Step 4 — Build feature matrices (X) and targets (y)
We convert each dataset into:
- **X**: the feature matrix (all descriptor columns)
- **y**: the solubility labels (`LogS`)

For AqSolDB descriptor files, we drop identifier columns (e.g., `Standardized_SMILES`) so that X contains only numerical features.

We also store the *exact feature column lists* from AqSolDB training data:
- `mord_cols` for Mordred
- `macc_cols` for MACCS

These column lists define the **training feature schema** and are used later to align Challenge 1/2 features so that models receive the same feature definitions at prediction time.


```python
def split_xy(df, target_col="LogS", id_col=None):
    y = df[target_col].astype(float).to_numpy()
    drop_cols = [target_col]
    if id_col is not None and id_col in df.columns:
        drop_cols.append(id_col)
    X = df.drop(columns=drop_cols)
    return X, y

# Train/Test feature matrices
Xtr_mord, ytr = split_xy(train_mord, TARGET_COL, ID_COL)
Xte_mord, yte = split_xy(test_mord,  TARGET_COL, ID_COL)

Xtr_macc, ytr2 = split_xy(train_macc, TARGET_COL, ID_COL)
Xte_macc, yte2 = split_xy(test_macc,  TARGET_COL, ID_COL)

assert len(ytr) == len(ytr2), "Train labels mismatch between Mordred and MACCS"
assert len(yte) == len(yte2), "Test labels mismatch between Mordred and MACCS"

mord_cols = Xtr_mord.columns.tolist()
macc_cols = Xtr_macc.columns.tolist()

# Challenge targets + SMILES
ch1_smiles = ch1[CH_SMILES_COL].astype(str).tolist()
ch1_y = ch1[TARGET_COL].astype(float).to_numpy()

ch2_smiles = ch2[CH_SMILES_COL].astype(str).tolist()
ch2_y = ch2[TARGET_COL].astype(float).to_numpy()

print("Mordred features:", len(mord_cols))
print("MACCS features:  ", len(macc_cols))
```

### Step 5 — Generate MACCS features for Challenge 1/2 and align to training schema
Challenge datasets provide SMILES, not precomputed feature matrices. Here we generate MACCS fingerprints from SMILES using RDKit.

Important detail: RDKit can produce a 167-position MACCS bit vector, while our course-provided AqSolDB MACCS files use a **fixed column schema** (e.g., 158 columns). To ensure consistency, we:
1. Infer a mapping from each dataset MACCS column to the most consistent RDKit bit index using a subset of the training molecules.
2. Apply this mapping to build Challenge MACCS features with the same column ordering as the training MACCS matrix.

We also track invalid SMILES (RDKit parsing failures) via a boolean mask so they can be excluded fairly during evaluation.

Reference: [RDKit MACCS keys](https://www.rdkit.org/new_docs/source/rdkit.Chem.MACCSkeys.html)


```python
# ===== MACCS for challenges (aligned to your training MACCS columns) =====

def rdkit_maccs_bits_167(smiles_list):
    X = np.zeros((len(smiles_list), 167), dtype=int)
    valid = np.ones(len(smiles_list), dtype=bool)
    for i, s in enumerate(smiles_list):
        mol = Chem.MolFromSmiles(s)
        if mol is None:
            valid[i] = False
            continue
        fp = MACCSkeys.GenMACCSKeys(mol)  # RDKit MACCS: 167 bits
        X[i, :] = [int(fp.GetBit(b)) for b in range(167)]
    return X, valid

def infer_maccs_mapping(train_df, smiles_col=ID_COL, prefix="MACCS_", sample_n=500):
    cols = [c for c in train_df.columns if c.startswith(prefix)]
    sample = train_df[[smiles_col] + cols].head(sample_n).copy()
    smiles = sample[smiles_col].astype(str).tolist()
    Y = sample[cols].to_numpy(dtype=int)

    X167, valid = rdkit_maccs_bits_167(smiles)
    X167 = X167[valid]
    Y = Y[valid]

    # match rate between each RDKit bit and each dataset MACCS column
    mean_match = (X167[:, :, None] == Y[:, None, :]).mean(axis=0)  # [167, n_cols]
    best_bit = mean_match.argmax(axis=0)
    best_score = mean_match.max(axis=0)

    mapping = {cols[i]: int(best_bit[i]) for i in range(len(cols))}
    print(f"[MACCS mapping] min match score across columns = {float(best_score.min()):.3f}")
    return mapping, cols

maccs_map, maccs_cols_order = infer_maccs_mapping(train_macc)

def make_maccs_df(smiles_list, mapping, cols_order):
    X167, valid = rdkit_maccs_bits_167(smiles_list)
    X = np.zeros((len(smiles_list), len(cols_order)), dtype=int)
    for j, col in enumerate(cols_order):
        X[:, j] = X167[:, mapping[col]]
    return pd.DataFrame(X, columns=cols_order), valid

Xch1_macc, valid1_macc = make_maccs_df(ch1_smiles, maccs_map, maccs_cols_order)
Xch2_macc, valid2_macc = make_maccs_df(ch2_smiles, maccs_map, maccs_cols_order)

print("Challenge1 MACCS:", Xch1_macc.shape, "invalid SMILES:", (~valid1_macc).sum())
print("Challenge2 MACCS:", Xch2_macc.shape, "invalid SMILES:", (~valid2_macc).sum())
```

### Step 6 — Compute Mordred descriptors for Challenge 1/2 and align to training schema
For the Mordred representation, we compute descriptors from the challenge SMILES using the Mordred calculator (`ignore_3D=True`).

Key points for ML correctness:
- We compute descriptors for Challenge 1/2, convert them to numeric values, and then **reindex** to the exact training feature list (`mord_cols`) so the model sees identical columns.
- Invalid SMILES are handled by inserting all-NaN rows at those positions and returning a validity mask. This keeps array shapes consistent while allowing correct filtering during evaluation.

Reference: [Mordred](https://doi.org/10.1186/s13321-018-0258-y)

```python
import numpy as np
import pandas as pd
from rdkit import Chem

# ===== Mordred for challenges (aligned to your training Mordred columns) =====
def compute_mordred_df(smiles_list, feature_cols):
    """
    Computes 2D Mordred descriptors and aligns them to the training feature columns.
    Robustly handles calculation errors by coercing them to NaN.
    """
    try:
        from mordred import Calculator, descriptors
    except ImportError as e:
        raise ImportError("Missing dependency 'mordred'. Install with: pip install mordred") from e

    # 1. Parse SMILES and identify valid molecules
    mols = []
    # Initialize mask as all False
    valid_mask = np.zeros(len(smiles_list), dtype=bool)
    
    for i, s in enumerate(smiles_list):
        # str(s) ensures we don't crash on non-string inputs like NaN
        m = Chem.MolFromSmiles(str(s))
        if m is not None:
            valid_mask[i] = True
            mols.append(m)
        else:
            # Invalid molecules are skipped in the list, but marked False in the mask
            pass

    # 2. Compute Descriptors
    calc = Calculator(descriptors, ignore_3D=True)
    
    if len(mols) > 0:
        # calc.pandas returns a DF that may contain 'MissingValue' objects or errors
        # quiet=True prevents spamming the console with progress bars if you prefer
        df_desc = calc.pandas(mols, quiet=True) 
        
        # --- CRITICAL FIX ---
        # 1. 'coerce' turns non-numeric error messages into NaN
        # 2. .astype(float) forces the entire block to be float64, preventing LossySetitemError
        df_desc = df_desc.apply(pd.to_numeric, errors='coerce').astype(float)
        
        # 3. Align with Feature Columns (Reindexing)
        # This discards extra columns and adds NaNs for missing ones
        df_desc = df_desc.reindex(columns=feature_cols)
    else:
        # Fallback if no valid molecules exist
        df_desc = pd.DataFrame(columns=feature_cols)

    # 3. Construct Final DataFrame
    # Prepare output container filled with NaNs
    out = pd.DataFrame(np.nan, index=np.arange(len(smiles_list)), columns=feature_cols)
    
    # Assign values into valid slots
    if len(mols) > 0:
        # Using .values ensures we insert the raw numpy array, avoiding index misalignment
        out.loc[valid_mask] = df_desc.values

    return out, valid_mask

# --- Usage ---
# Ensure 'mord_cols' is defined (from your training step, e.g., X_train.columns)
# If you named it something else (like mordred_feature_cols), swap that variable in below.

print("Computing Mordred for Challenge 1...")
Xch1_mord, valid1_mord = compute_mordred_df(ch1_smiles, mord_cols)

print("Computing Mordred for Challenge 2...")
Xch2_mord, valid2_mord = compute_mordred_df(ch2_smiles, mord_cols)

print("Challenge1 Mordred:", Xch1_mord.shape, "invalid SMILES:", (~valid1_mord).sum())
print("Challenge2 Mordred:", Xch2_mord.shape, "invalid SMILES:", (~valid2_mord).sum())
```

### Step 7 — Cross-validation and hyperparameter tuning (SVR and RF)
We tune model hyperparameters using **5-fold cross-validation** on the **AqSolDB training split only**.

What this cell does:
- Defines the evaluation metrics used in this project:
  - RMSE, MAE, R², and CCC (Concordance Correlation Coefficient).
- Builds **pipelines** so preprocessing is learned from training data only (prevents leakage):
  - Mordred + SVR: median imputation + scaling + SVR(RBF)
  - Mordred + RF: median imputation + Random Forest
  - MACCS + SVR: scaling + SVR(RBF)
  - MACCS + RF: Random Forest
- Runs **GridSearchCV** for each model/representation pair using RMSE as the selection criterion (`neg_root_mean_squared_error`).
- Prints the best cross-validated RMSE and the corresponding best hyperparameters.

References: [scikit-learn Pipeline](https://scikit-learn.org/stable/modules/compose.html#pipeline), [GridSearchCV](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GridSearchCV.html), [CCC](https://doi.org/10.2307/2532051), [SVR](https://doi.org/10.1023/B:STCO.0000035301.49549.88), [Random Forests](https://doi.org/10.1023/A:1010933404324)


```python
# ===== Cross-validation + GridSearchCV hyperparameter tuning =====

def ccc(y_true, y_pred):
    """Lin's concordance correlation coefficient (CCC)."""
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    mu_x = y_true.mean()
    mu_y = y_pred.mean()
    var_x = y_true.var()
    var_y = y_pred.var()
    if var_x == 0 or var_y == 0:
        return np.nan
    cov = np.mean((y_true - mu_x) * (y_pred - mu_y))
    return float((2 * cov) / (var_x + var_y + (mu_x - mu_y) ** 2))

def metric_dict(y_true, y_pred):
    rmse = float(np.sqrt(mean_squared_error(y_true, y_pred)))
    mae  = float(mean_absolute_error(y_true, y_pred))
    r2   = float(r2_score(y_true, y_pred))
    ccc_v = ccc(y_true, y_pred)
    return {"RMSE": rmse, "MAE": mae, "R2": r2, "CCC": ccc_v}

cv = KFold(n_splits=5, shuffle=True, random_state=42)
scoring = "neg_root_mean_squared_error"

# Pipelines
pipe_svr_mord = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("svr", SVR(kernel="rbf"))
])

pipe_rf_mord = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("rf", RandomForestRegressor(random_state=42, n_jobs=-1))
])

pipe_svr_macc = Pipeline([
    ("scaler", StandardScaler()),  # you used scaling for MACCS+SVR
    ("svr", SVR(kernel="rbf"))
])

pipe_rf_macc = Pipeline([
    ("rf", RandomForestRegressor(random_state=42, n_jobs=-1))
])

# Parameter grids (explainable and not too large)
grid_svr_mord = {
    "svr__C": [0.1, 1, 10, 100],
    "svr__epsilon": [0.01, 0.05, 0.1, 0.2],
    "svr__gamma": ["scale"]
}

grid_rf_mord = {
    "rf__n_estimators": [300, 600],
    "rf__max_depth": [None, 20, 40],
    "rf__min_samples_leaf": [1, 2, 4],
    "rf__max_features": ["sqrt", "log2"]
}

grid_svr_macc = {
    "svr__C": [0.1, 1, 10, 100],
    "svr__epsilon": [0.01, 0.05, 0.1, 0.2],
    "svr__gamma": ["scale", 0.01, 0.1]
}

grid_rf_macc = {
    "rf__n_estimators": [300, 600, 900],
    "rf__max_depth": [None, 10, 20, 40],
    "rf__min_samples_leaf": [1, 2, 4]
}

def fit_gs(pipe, grid, X, y, name):
    gs = GridSearchCV(pipe, grid, cv=cv, scoring=scoring, n_jobs=-1, verbose=1)
    gs.fit(X, y)
    best_rmse = -float(gs.best_score_)
    print(f"\n[{name}] Best CV RMSE: {best_rmse:.4f}")
    print(f"[{name}] Best params: {gs.best_params_}")
    return gs.best_estimator_, best_rmse, gs.best_params_

best_svr_mord, cv_rmse_svr_mord, params_svr_mord = fit_gs(pipe_svr_mord, grid_svr_mord, Xtr_mord, ytr, "Mordred-SVR")
best_rf_mord,  cv_rmse_rf_mord,  params_rf_mord  = fit_gs(pipe_rf_mord,  grid_rf_mord,  Xtr_mord, ytr, "Mordred-RF")

best_svr_macc, cv_rmse_svr_macc, params_svr_macc = fit_gs(pipe_svr_macc, grid_svr_macc, Xtr_macc, ytr, "MACCS-SVR")
best_rf_macc,  cv_rmse_rf_macc,  params_rf_macc  = fit_gs(pipe_rf_macc,  grid_rf_macc,  Xtr_macc, ytr, "MACCS-RF")
```

# 5 Evalulation

### 1. RMSE (Root Mean Square Error)
**Measures:** Average magnitude of prediction error (lower is better).
$$RMSE = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2}$$
* $y_i$: Actual value
* $\hat{y}_i$: Predicted value

### 2. $R^2$ (Coefficient of Determination)
**Measures:** Proportion of variance explained. 1.0 is perfect; negative values indicate the model is worse than guessing the average.
$$R^2 = 1 - \frac{\sum (y_i - \hat{y}_i)^2}{\sum (y_i - \bar{y})^2}$$
* $\bar{y}$: Mean of actual values

### 3. CCC (Concordance Correlation Coefficient)
**Measures:** Agreement between predicted and actual values. Unlike standard correlation, it penalizes predictions that are shifted or scaled incorrectly (checks adherence to $y=x$).
$$CCC = \frac{2 \rho \sigma_{true} \sigma_{pred}}{\sigma_{true}^2 + \sigma_{pred}^2 + (\mu_{true} - \mu_{pred})^2}$$
* $\rho$: Pearson correlation coefficient
* $\mu, \sigma$: Mean and variance of true/predicted values

## 5.1. Premise: The "Aleatoric Limit"
In QSAR modeling, an RMSE of 0.7–0.9 log units is generally considered the gold standard. This is because the experimental error in solubility measurements is approximately 0.6 log units, meaning that errors lower than this usually indicate that the model is capturing noise rather than genuine signal (Palmer et al., 2014).
A recent study explicitly labels this range as the "aleatoric limit" (0.5–1.0 logS), highlighting that predictions below this threshold are essentially impossible due to the irreducible error inherent in the training data (Attia et al., 2024).
As a result, for large-scale datasets, state-of-the-art consensus models generally reach a plateau at an $R^2$ of 0.87–0.89, which represents a practical ceiling for true predictive performance (Sorkun et al., 2020). For rigorous validation, the Concordance Correlation Coefficient (CCC) is also considered, with values above 0.85 being expected. CCC is particularly useful because it evaluates both correlation and bias in the predictions (Chirico and Gramatica, 2011).
## 5.2. Current Evaluation State
To determine whether our models approach this aleatoric limit, we adopt a multitiered evaluation framework. Our evaluation is structured as a hierarchy of metrics applied to both the Test Set (interpolation) and Challenge Sets (extrapolation).
- RMSE & MAE (Error Magnitude): Root Mean Square Error (RMSE) serves as the primary benchmark, compared against the experimental noise floor (~0.6 log units). Mean Absolute Error (MAE) provides a complementary measure of average prediction error.
- $R^2$ & Pearson (Variance & Ranking): $R^2$ measures the proportion of variance explained. Pearson correlation is paired with $R^2$ to check whether the model correctly ranks molecular solubility even when absolute predictions are imperfect.
- CCC (Calibration Check): The Concordance Correlation Coefficient (CCC) is included to detect miscalibration. Unlike standard correlation, CCC penalizes predictions that deviate from the ideal $y=x$ line, revealing cases where a model correctly ranks molecules but predicts the wrong scale.

```python
# ===== Evaluate best CV-selected models on Test + Challenge sets =====

rows = []

def eval_add(rep, model_name, model, X, y, dataset, valid_mask=None):
    y_pred = model.predict(X)
    if valid_mask is not None:
        y_true = y[valid_mask]
        y_pred = y_pred[valid_mask]
    else:
        y_true = y
    rows.append({
        "Dataset": dataset,
        "Representation": rep,
        "Model": model_name,
        **metric_dict(y_true, y_pred)
    })

# AqSolDB test
eval_add("Mordred", "SVR", best_svr_mord, Xte_mord, yte, "Test")
eval_add("Mordred", "RF",  best_rf_mord,  Xte_mord, yte, "Test")
eval_add("MACCS",   "SVR", best_svr_macc, Xte_macc, yte2, "Test")
eval_add("MACCS",   "RF",  best_rf_macc,  Xte_macc, yte2, "Test")

# Challenge 1
eval_add("Mordred", "SVR", best_svr_mord, Xch1_mord, ch1_y, "Challenge1", valid_mask=valid1_mord)
eval_add("Mordred", "RF",  best_rf_mord,  Xch1_mord, ch1_y, "Challenge1", valid_mask=valid1_mord)
eval_add("MACCS",   "SVR", best_svr_macc, Xch1_macc, ch1_y, "Challenge1", valid_mask=valid1_macc)
eval_add("MACCS",   "RF",  best_rf_macc,  Xch1_macc, ch1_y, "Challenge1", valid_mask=valid1_macc)

# Challenge 2
eval_add("Mordred", "SVR", best_svr_mord, Xch2_mord, ch2_y, "Challenge2", valid_mask=valid2_mord)
eval_add("Mordred", "RF",  best_rf_mord,  Xch2_mord, ch2_y, "Challenge2", valid_mask=valid2_mord)
eval_add("MACCS",   "SVR", best_svr_macc, Xch2_macc, ch2_y, "Challenge2", valid_mask=valid2_macc)
eval_add("MACCS",   "RF",  best_rf_macc,  Xch2_macc, ch2_y, "Challenge2", valid_mask=valid2_macc)

results_df = pd.DataFrame(rows).sort_values(["Dataset", "Representation", "Model"])
display(results_df)

# Save results
results_df.to_csv("cv_tuned_metrics_test_ch1_ch2_with_ccc.csv", index=False)
print("Saved: cv_tuned_metrics_test_ch1_ch2_with_ccc.csv")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.manifold import TSNE
import umap
from mpl_toolkits.mplot3d import Axes3D

# 1. Check for data leakage (SMILES overlap)
train_smiles_set = set(train_mord[ID_COL])
test_smiles_set = set(test_mord[ID_COL])
ch1_smiles_set = set(ch1_smiles)
ch2_smiles_set = set(ch2_smiles)

overlap_ch1 = train_smiles_set.intersection(ch1_smiles_set)
overlap_ch2 = train_smiles_set.intersection(ch2_smiles_set)
print(f"Overlap Train & Challenge 1: {len(overlap_ch1)}")
print(f"Overlap Train & Challenge 2: {len(overlap_ch2)}")

# 2. Prepare feature matrices for PCA/t-SNE/UMAP
X_train_clean = Xtr_mord.dropna(how='all'); y_train_labels = ['AqSolDB (Train)']*len(X_train_clean)
X_test_clean = Xte_mord.dropna(how='all'); y_test_labels = ['AqSolDB (Test)']*len(X_test_clean)
X_ch1_clean = Xch1_mord.dropna(how='all'); y_ch1_labels = ['Challenge 1']*len(X_ch1_clean)
X_ch2_clean = Xch2_mord.dropna(how='all'); y_ch2_labels = ['Challenge 2']*len(X_ch2_clean)

X_all = pd.concat([X_train_clean, X_test_clean, X_ch1_clean, X_ch2_clean])
y_all = y_train_labels + y_test_labels + y_ch1_labels + y_ch2_labels

# 3. Impute and scale features
X_imputed = SimpleImputer(strategy='median').fit_transform(X_all)
X_scaled = StandardScaler().fit_transform(X_imputed)

# --- 2D PCA ---
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
df_pca = pd.DataFrame(X_pca, columns=['PC1','PC2']); df_pca['Dataset'] = y_all
plt.figure(figsize=(12,8))
sns.scatterplot(data=df_pca[df_pca['Dataset'].str.contains('AqSolDB')], x='PC1', y='PC2', hue='Dataset', palette=['#a1c9f4','#8de5a1'], alpha=0.2, s=30, linewidth=0)
sns.scatterplot(data=df_pca[~df_pca['Dataset'].str.contains('AqSolDB')], x='PC1', y='PC2', hue='Dataset', palette=['#ff9f4b','#ff4b4b'], alpha=0.9, s=60, edgecolor='black')
plt.title("Chemical Space Analysis (PCA)")
plt.xlabel(f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}% var)")
plt.ylabel(f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}% var)")
plt.legend(bbox_to_anchor=(1.05,1), loc=2)
plt.tight_layout()
plt.show()

# --- 2D t-SNE ---
tsne = TSNE(n_components=2, perplexity=30, random_state=42, n_jobs=-1)
X_tsne = tsne.fit_transform(X_scaled)
df_tsne = pd.DataFrame(X_tsne, columns=['Dim1','Dim2']); df_tsne['Dataset'] = y_all
plt.figure(figsize=(12,8))
sns.scatterplot(data=df_tsne[df_tsne['Dataset'].str.contains('AqSolDB')], x='Dim1', y='Dim2', hue='Dataset', palette=['#a1c9f4','#8de5a1'], alpha=0.2, s=30, linewidth=0)
sns.scatterplot(data=df_tsne[~df_tsne['Dataset'].str.contains('AqSolDB')], x='Dim1', y='Dim2', hue='Dataset', palette=['#ff9f4b','#ff4b4b'], alpha=0.8, s=50, edgecolor='black', linewidth=0.5)
plt.title("Chemical Space Analysis (t-SNE)")
plt.xlabel("t-SNE Dim 1"); plt.ylabel("t-SNE Dim 2")
plt.legend(bbox_to_anchor=(1.05,1), loc=2)
plt.tight_layout()
plt.show()

# --- 2D UMAP ---
reducer = umap.UMAP(n_components=2, n_neighbors=30, min_dist=0.3, random_state=42, n_jobs=-1)
X_umap = reducer.fit_transform(X_scaled)
df_umap = pd.DataFrame(X_umap, columns=['UMAP1','UMAP2']); df_umap['Dataset'] = y_all
plt.figure(figsize=(12,8))
sns.scatterplot(data=df_umap[df_umap['Dataset'].str.contains('AqSolDB')], x='UMAP1', y='UMAP2', hue='Dataset', palette=['#a1c9f4','#8de5a1'], alpha=0.2, s=30, linewidth=0)
sns.scatterplot(data=df_umap[~df_umap['Dataset'].str.contains('AqSolDB')], x='UMAP1', y='UMAP2', hue='Dataset', palette=['#ff9f4b','#ff4b4b'], alpha=0.8, s=40, edgecolor='black', linewidth=0.3)
plt.title("Chemical Space Analysis (UMAP)")
plt.xlabel("UMAP Dim 1"); plt.ylabel("UMAP Dim 2")
plt.legend(bbox_to_anchor=(1.05,1), loc=2)
plt.tight_layout()
plt.show()

# --- 3D Projections ---
pca_3d = PCA(n_components=3).fit_transform(X_scaled)
tsne_3d = TSNE(n_components=3, perplexity=30, random_state=42, n_jobs=-1).fit_transform(X_scaled)
X_umap_3d = umap.UMAP(n_components=3, n_neighbors=30, min_dist=0.3, random_state=42, n_jobs=-1).fit_transform(X_scaled)

fig = plt.figure(figsize=(24,8))

def plot_3d_subplot(ax, data, title):
    df_temp = pd.DataFrame(data, columns=['x','y','z']); df_temp['Dataset'] = y_all
    ax.scatter(df_temp[df_temp['Dataset'].str.contains('AqSolDB')]['x'], df_temp[df_temp['Dataset'].str.contains('AqSolDB')]['y'], df_temp[df_temp['Dataset'].str.contains('AqSolDB')]['z'], c='#a1c9f4', alpha=0.1, s=10, label='AqSolDB')
    ax.scatter(df_temp[df_temp['Dataset']=='Challenge 1']['x'], df_temp[df_temp['Dataset']=='Challenge 1']['y'], df_temp[df_temp['Dataset']=='Challenge 1']['z'], c='#ff9f4b', alpha=0.6, s=20, label='Challenge 1')
    ax.scatter(df_temp[df_temp['Dataset']=='Challenge 2']['x'], df_temp[df_temp['Dataset']=='Challenge 2']['y'], df_temp[df_temp['Dataset']=='Challenge 2']['z'], c='#ff4b4b', alpha=0.8, s=25, label='Challenge 2')
    ax.set_title(title, fontsize=15, fontweight='bold')
    ax.set_xlabel('Dim 1'); ax.set_ylabel('Dim 2'); ax.set_zlabel('Dim 3'); ax.view_init(elev=30, azim=135)

ax1 = fig.add_subplot(131, projection='3d'); plot_3d_subplot(ax1, pca_3d, f"Linear: 3D PCA")
ax2 = fig.add_subplot(132, projection='3d'); plot_3d_subplot(ax2, tsne_3d, "Local Clustering: 3D t-SNE")
ax3 = fig.add_subplot(133, projection='3d'); plot_3d_subplot(ax3, X_umap_3d, "Global Structure: 3D UMAP")
plt.legend(bbox_to_anchor=(1.05,1), loc=2, markerscale=2)
plt.tight_layout()
plt.show()
```

## 5.4. Extrapolation Challenges

Based on the error analysis and chemical property plots, the largest prediction errors occur on non-drug-like compounds, such as organometallic or inorganic complexes (e.g., Zirconium, Iron), where 2D descriptors miss coordination chemistry and lattice effects. For bulky hydrophobic polysulfides, MACCS keys underestimate size, whereas Mordred descriptors capture steric bulk and lipophilicity more effectively.

### Extrapolation Hypothesis

**Challenge 1 (Orange): The Big Molecules**
Molecules larger than 500 Da show near-normal solubility. Mordred descriptors are expected to perform better than MACCS, as detailed features are required to distinguish subtle structural differences. MACCS is too simple for complex structures.

**Challenge 2 (Green): The Super Insoluble Molecules**
Molecules with extremely low solubility (-7 LogS), far outside the model's training domain. All models (RF and SVR) are expected to fail, producing negative R² and high RMSE.

UMAP and t-SNE plots show that Challenge 2 **(RED)** forms a distinct cluster, confirming that these molecules represent a true extrapolation scenario and explain the major drop in model performance.

```python
import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import Descriptors, Draw

# 1. Analyze failures for forensic evaluation
def analyze_failures(X_mord, X_macc, y_true, smiles_list, dataset_name):
    df_temp = pd.DataFrame({'SMILES': smiles_list, 'Actual_LogS': y_true})
    try:
        df_temp['Pred_Mord'] = best_rf_mord.predict(X_mord)
    except:
        df_temp['Pred_Mord'] = np.nan
    try:
        df_temp['Pred_MACCS'] = best_rf_macc.predict(X_macc)
    except:
        df_temp['Pred_MACCS'] = np.nan
    df = df_temp.dropna().copy()
    df['Err_Mord'] = np.abs(df['Actual_LogS'] - df['Pred_Mord'])
    df['Err_MACCS'] = np.abs(df['Actual_LogS'] - df['Pred_MACCS'])
    df['Mean_Error'] = (df['Err_Mord'] + df['Err_MACCS']) / 2
    df['MW'] = [Descriptors.MolWt(Chem.MolFromSmiles(s)) if Chem.MolFromSmiles(s) else 0 for s in df['SMILES']]
    df['LogP'] = [Descriptors.MolLogP(Chem.MolFromSmiles(s)) if Chem.MolFromSmiles(s) else 0 for s in df['SMILES']]
    return df

# Generate rogue analysis for Challenge sets
df_ch1_rogues = analyze_failures(Xch1_mord, Xch1_macc, ch1_y, ch1_smiles, "Challenge 1")
df_ch2_rogues = analyze_failures(Xch2_mord, Xch2_macc, ch2_y, ch2_smiles, "Challenge 2")

# 2. Visualize top failures
def show_top_rogues(df, title):
    top_rogues = df.sort_values('Mean_Error', ascending=False).head(5)
    print(f"\n=== {title} : TOP 5 FAILURES ===")
    for i, (_, row) in enumerate(top_rogues.iterrows(), 1):
        print(f"#{i} | LogS: {row['Actual_LogS']:.2f} | MW: {row['MW']:.0f} | LogP: {row['LogP']:.1f}")
        print(f"    Mord Error: {row['Err_Mord']:.2f} | MACCS Error: {row['Err_MACCS']:.2f}")
        print(f"    SMILES: {row['SMILES']}")
        print("-"*40)
    mols = [Chem.MolFromSmiles(s) for s in top_rogues['SMILES']]
    legends = [f"True: {row['Actual_LogS']:.1f}\nErr: {row['Mean_Error']:.1f}\nMW: {row['MW']:.0f}" for _, row in top_rogues.iterrows()]
    img = Draw.MolsToGridImage(mols, molsPerRow=5, subImgSize=(250,250), legends=legends)
    display(img)

# Display top rogues for each Challenge
show_top_rogues(df_ch1_rogues, "CHALLENGE 1 (The Heavyweights)")
show_top_rogues(df_ch2_rogues, "CHALLENGE 2 (The Insoluble Wall)")
```

## 5.5 Conclusion: Why the Model Got Confused
The model didn’t fail due to lack of data, but because it learned from the wrong type of examples and only saw molecules in 2D.
The training data (AqSolDB) mostly contains small, drug-like, soluble molecules, so the model wasn’t prepared for the unusual and extreme molecules in the challenge datasets.
In Challenge 2, the model could count rings but didn’t understand that flat molecules can stack tightly like bricks, making them essentially insoluble.
In Challenge 1, the model saw polar groups and flexibility but didn’t realize that large molecules can fold and hide their polar regions from water, affecting solubility.

## 5.6. Limitations & Potential Extensions (Data & Features)

Our evaluation provides valuable insights into solubility prediction, but several areas could further strengthen the analysis if more data or advanced features were available:

- **Broader & More Diverse Training Data:** Although we included many drug-like molecules and calculated additional physicochemical properties, the training set still underrepresents extremely large, highly insoluble, or unusual molecules. Expanding the dataset could reduce extrapolation failures observed in Challenge 1 & 2.

- **3D Structural Features:** We currently rely on 2D descriptors (Mordred) and fingerprints (MACCS). Incorporating 3D descriptors—such as steric bulk, molecular volume, hydrogen bonding patterns, or stacking potential—could improve predictions for molecules where folding or planarity strongly affects solubility.

- **Advanced Feature Analysis:** While we analyzed correlations and feature-target relationships, deeper exploration (e.g., feature importance ranking across models or non-linear interactions) could reveal which descriptors fail for extreme molecules, guiding better feature selection or engineering.

- **Cross-Representation Integration:** Using additional descriptor sets (e.g., ECFP, topological or electronic descriptors) could provide complementary perspectives, helping the model capture subtle chemical effects not represented in 2D features alone.

- **Out-of-Distribution Benchmarking:** Larger or more chemically diverse challenge sets would allow a more robust evaluation of the model’s extrapolation limits, helping quantify uncertainty for unusual molecules.

## 5.7. Quantitative Observations
- Test Set (In-Distribution): The model achieved an RMSE of 0.97 and CCC of 0.90. This confirms that the model successfully learned the physics of the training domain (AqSolDB) and generalizes well to similar chemistry.

- Challenge 1 (Feature Shift): The drop in ranking ability (R2 -0.08) correlates with the shift toward higher molecular weights. The model likely struggles here because the training data under-represents these complex folds ("The Heavyweights"), forcing it to extrapolate beyond its experience.

Challenge 2 (Distribution Shift): The catastrophic metrics (RMSE 3.58, R2 -24.47). As seen in the UMAP plots, these compounds occupy a region of chemical space completely absent from the training set. The model is failing because it is being asked to predict a target range (extreme insolubility) it has never seen before.
